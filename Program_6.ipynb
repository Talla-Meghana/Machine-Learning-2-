```python
import random
def uniform_random():
  return random.uniform(0,1)
def f(x):
  return x**2
n=10
samples=[uniform_random() for __ in range(n)]
f_values=[f(x) for x in samples]
print('random samples:',samples)
print('functional values:',f_values)
i_est=sum(f_values)/n
print('estimated value of monte carlo estimatation:',i_est)
i_true=1/3
print('original value of monte carlo estimation:',i_true)
```
